# Como instalar/configurar/usar o `tumbler` no `Linux Ubuntu`

## Resumo

Guia para instalar o `tumbler` pelo `apt` no `Linux Ubuntu`. O `tumbler` é um serviço D-Bus que gera miniaturas de arquivos para ambientes gráficos, como o Xfce.

## _Abstract_

_A guide to install `tumbler` through `apt` on `Linux Ubuntu`. `Tumbler` is a D-Bus thumbnailing service for desktop environments such as Xfce._


## Descrição

### `tumbler`

O `tumbler` é um serviço D-Bus para gerar miniaturas de arquivos. Ambientes gráficos, como o Xfce, podem usá-lo para mostrar prévias de imagens, documentos e outros formatos compatíveis no gerenciador de arquivos.


## Pré-requisitos

- Linux Ubuntu com repositório `universe` habilitado
- Permissão para usar `sudo`
- Conexão com a internet para baixar o pacote
- Gerenciador de pacotes `apt` funcional


## 1. Abrir o `Terminal Emulator`

1. Abrir o `Terminal Emulator`. Você pode fazer isso pressionando:

    ```bash
    Ctrl + Alt + T
    ```


2. Certifique-se de que seu sistema esteja limpo e atualizado.

    2.1 Limpar o `cache` do gerenciador de pacotes `apt`. Especificamente, ele remove todos os arquivos de pacotes (`.deb`) baixados pelo `apt` e armazenados em `/var/cache/apt/archives/`. Digite o seguinte comando:

    ```bash
    sudo apt clean
    ```

    2.2 Remover pacotes `.deb` antigos ou duplicados do `cache` local. É útil para liberar espaço, pois remove apenas os pacotes que não podem mais ser baixados (ou seja, versões antigas de pacotes que foram atualizados). Digite o seguinte comando:

    ```bash
    sudo apt autoclean
    ```

    2.3 Remover pacotes que foram automaticamente instalados para satisfazer as dependências de outros pacotes e que não são mais necessários. Digite o seguinte comando:

    ```bash
    sudo apt autoremove -y
    ```

    2.4 Buscar as atualizações disponíveis para os pacotes que estão instalados em seu sistema. Digite o seguinte comando e pressione `Enter`:

    ```bash
    sudo apt update
    ```

    2.5 **Corrigir pacotes quebrados**: Isso atualizará a lista de pacotes disponíveis e tentará corrigir pacotes quebrados ou com dependências ausentes:

    ```bash
    sudo apt --fix-broken install
    ```

    2.6 Limpar o `cache` do gerenciador de pacotes `apt` novamente:

    ```bash
    sudo apt clean
    ```

    2.7 Para ver a lista de pacotes a serem atualizados, digite o seguinte comando e pressione `Enter`:

    ```bash
    sudo apt list --upgradable
    ```

    2.8 Realmente atualizar os pacotes instalados para as suas versões mais recentes, com base na última vez que você executou `sudo apt update`. Digite o seguinte comando e pressione `Enter`:

    ```bash
    sudo apt full-upgrade -y
    ```


## 3. Instalar o `tumbler` via `apt`

O pacote está disponível no repositório `universe` do Ubuntu. O `apt` instala também as dependências necessárias automaticamente.

1. Habilitar o repositório `universe` e atualizar a lista de pacotes:

    ```bash
    sudo add-apt-repository universe
    sudo apt update
    ```

2. Instalar o pacote:

    ```bash
    sudo apt install -y tumbler
    ```

3. Confirmar que o pacote está instalado:

    ```bash
    dpkg-query -W -f='${Status} ${Version}\n' tumbler
    ```


## 4. Usar o `tumbler`

O `tumbler` é iniciado sob demanda por clientes compatíveis via D-Bus, como o gerenciador de arquivos do Xfce. Em geral, não é necessário iniciar o serviço manualmente.

1. Em um ambiente Xfce, abra o gerenciador de arquivos e navegue até uma pasta com imagens ou documentos para que as miniaturas sejam geradas.

2. Para consultar os arquivos instalados pelo pacote, use:

    ```bash
    dpkg -L tumbler
    ```

3. Para consultar informações sobre o pacote e suas dependências:

    ```bash
    apt show tumbler
    ```


## Compatibilidade

- O pacote `tumbler` é distribuído pelo repositório `universe` do Ubuntu.
- A disponibilidade e a versão dependem da versão do Ubuntu e da arquitetura do sistema.
- Recursos de miniaturas dependem dos plugins instalados e dos aplicativos que solicitam as prévias.
